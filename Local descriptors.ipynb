```python
import pandas as pd
from mendeleev import element

# Load the Excel file
file_path = r"C:\Users\Lenovo\Desktop\ML-亲电性\训练数据表\OutofSamples.xlsx"  # Replace with your actual Excel file path
sheet_name = 'Metal'          # Replace with your sheet name if necessary
# Read specific columns (e.g., 'C', 'N', 'C') from the Excel file
df = pd.read_excel(file_path, sheet_name=sheet_name, usecols=['Atom'])
#print(df)
file = df['Atom'].tolist() 
print(file)


for M in file:
    #print(X)
    X=element(M)
    #print(X)
    #MAR=X.atomic_number
    #print(X.atomic_number)
    print('{M} {AR} {AN} {AW} {EA} {DP} {EV} {EN}'.format(M=M, AR=X.atomic_radius, AN=X.atomic_number, AW=X.atomic_weight, EA=X.electron_affinity, DP=X.dipole_polarizability, EV=X.nvalence(), EN=X.electronegativity(scale="pauling")))

   
   
```
