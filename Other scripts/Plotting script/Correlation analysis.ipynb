```python
import matplotlib.pyplot as plt
import pandas as pd

# 1. Read the Excel file (replace the file path with your actual path)
df = pd.read_excel(r"C:\Users\Lenovo\Desktop\NBO自旋-亲核性相关性分析.xlsx")

# 2. Extract the two columns of data for comparison 
x = df["f"]
y = df["NBO_Spin"]

# 3. Start plotting
plt.figure(figsize=(6, 6))  # Set canvas size to a square

# Plot scatter points
plt.scatter(x, y, color="royalblue", alpha=0.7, edgecolors="k", label="Molecules")

# (Optional) Draw a y = x reference line to easily observe deviations
min_val = min(min(x), min(y))
max_val = max(max(x), max(y))
plt.plot(
    [min_val, max_val],
    [min_val, max_val],
    color="gray",
    linestyle="--",
    label="Ideal y=x"
)

# Axis labels
plt.xlabel("Nucleophilicity of Radical Center", fontsize=12)
plt.ylabel("NPA_spin", fontsize=12)
#plt.title("Gas-phase vs. SMD(DCM) Comparison", fontsize=14)

# Add legend and grid
plt.legend()
plt.grid(True, linestyle=":", alpha=0.6)

# 4. Save and show the plot
plt.tight_layout()
plt.savefig(
    "gas_vs_smd_comparison.png", dpi=300
)  # Save as a high-resolution image
plt.show()
```
