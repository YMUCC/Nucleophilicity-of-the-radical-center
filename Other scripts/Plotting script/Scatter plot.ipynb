```python
import numpy as np
import matplotlib.pyplot as plt

# Create a 1x1 figure layout
fig, axs = plt.subplots(nrows=1, ncols=1, figsize=(9, 8))

# Plot scatter plots
axs.scatter(data['xgb_train'], data['y_train'], s=120, color='#1D3557', alpha=0.75) 
axs.scatter(data['xgb_test'], data['y_test'], s=120, color='#E63946', alpha=0.85)
#axs.scatter(data['lgbm_train'], data['y_train'], s=120, color='#5D3A9B', alpha=0.8)
#axs.scatter(data['lgbm_test'], data['y_test'], s=120, color='#E69F00', alpha=0.9)
#axs.scatter(data['rf_train'], data['y_train'], s=120, color='b')
#axs.scatter(data['rf_test'], data['y_test'], s=120, color='#64B5F6')

#axs.set_title('RF', fontsize=18)
axs.axline((0, 0), slope=1, color='gray', linestyle='--', alpha=0.5, linewidth=1)
axs.set_xlabel('XGB Predicted Value', fontsize=24, fontweight='semibold')
axs.set_ylabel('DFT Calculated Value', fontsize=24, fontweight='semibold')

axs.grid(True, alpha=0.3, linestyle='--', linewidth=0.5)
axs.tick_params(axis='both', which='major', labelsize=24)
axs.tick_params(axis='both', which='minor', labelsize=22)

axs.legend(fontsize=12, 
           frameon=True, 
           fancybox=True, 
           shadow=True,
           borderpad=1,
           labelspacing=1)
plt.style.use('seaborn-v0_8-whitegrid')  # Use seaborn whitegrid style

fig.tight_layout()
fig.subplots_adjust(hspace=0.26, wspace=0.02)

# ================= Modify here: Embed metrics directly into legend labels =================
# Replace variables with your actual loaded data; format matches the target layout
label_train = f'Train (MAE={0.006}, $R^2$ = {0.995})'
label_test = f'Test (MAE={0.058}, $R^2$ = {0.82})'

axs.legend([label_train, label_test], fontsize=20, handlelength=2, frameon=True, handletextpad=0.01)
# ====================================================================================

plt.show()
```
